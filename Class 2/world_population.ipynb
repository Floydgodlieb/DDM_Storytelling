```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

df = pd.read_csv("world_population.csv")

# scale marker size (sqrt so area ~ population, not radius)
marker_size = (df["2022 Population"]**0.5)/30

# log-transform population for color, so small countries aren't all one color
log_pop = np.log10(df["2022 Population"])

fig = go.Figure(go.Scattergeo(
    locations=df["CCA3"],
    text=df["Country/Territory"],
    customdata=df["2022 Population"],
    marker=dict(
        size=marker_size,
        sizemode="area",
        color=log_pop,
        colorscale="Viridis",
        colorbar=dict(
            title="Population",
            tickvals=[6, 7, 8, 9],
            ticktext=["1M", "10M", "100M", "1B"],
        ),
        line=dict(width=0.5, color="white"),
    ),
    hovertemplate=(
        "<b>%{text}</b><br>"
        "Population 2022: %{customdata:,.0f}<br>"
        "<extra></extra>"
    ),
))

fig.update_geos(
    projection_type="orthographic",
    showcountries=True,
    showland=True,
    landcolor="lightgray",
    showocean=True,
    oceancolor="lightblue",
)

fig.update_layout(height=800, margin=dict(l=0, r=0, t=0, b=0))
fig.show()
```
